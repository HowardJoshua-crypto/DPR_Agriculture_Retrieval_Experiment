```python
from sentence_transformers import SentenceTransformer
import numpy as np
import torch

MODEL_ID="sentence-transformers/all-MiniLM-L6-v2"
model= SentenceTransformer(MODEL_ID)
print("Model: ", MODEL_ID)
print("Device: ",model.device)
```

**Dense Retrieval** converts each passage into a numerical ventor

```python
passage_texts=corpus_df["text"].tolist()

passage_embeddings=model.encode(
    passage_texts,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("Embedding shape: ",passage_embeddings.shape)
print("First passage vector:")
print(passage_embeddings[0][:10])

```

```python

def dense_search(question, top_k=5):
    query_embedding = model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )[0]

    # Normalized vectors make dot product equivalent
    # to cosine similarity.
    scores = passage_embeddings @ query_embedding

    ranked_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for rank, i in enumerate(ranked_indices, start=1):
        results.append({
            "rank": rank,
            "passage_id": corpus_df.iloc[i]["passage_id"],
            "score": float(scores[i]),
            "text": corpus_df.iloc[i]["text"]
        })

    return pd.DataFrame(results)


question = "How can farmers improve nitrogen availability in soil?"

dense_results = dense_search(question)

display(dense_results)


```
